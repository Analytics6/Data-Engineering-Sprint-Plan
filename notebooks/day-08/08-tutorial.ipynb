```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AzureDE-InterviewPrep')
    .master('local[*]')
    .config('spark.sql.shuffle.partitions', '4')
    .getOrCreate()
)
spark
```

```python
spark.createDataFrame(
    [(1, 'Alice', 10), (2, 'Bob', 20), (3, 'Carol', 10), (4, 'Dan', None)],
    ['emp_id', 'name', 'dept_id'],
).createOrReplaceTempView('employees')
spark.createDataFrame(
    [(10, 'Sales'), (20, 'Eng'), (30, 'HR')],
    ['dept_id', 'dept_name'],
).createOrReplaceTempView('departments')
spark.createDataFrame(
    [(1, 100), (1, 200), (2, 50), (4, 80)],
    ['emp_id', 'amount'],
).createOrReplaceTempView('sales')
```

```python
spark.sql('''
SELECT e.name, d.dept_name, s.amount
FROM employees e
LEFT JOIN departments d ON e.dept_id = d.dept_id
LEFT JOIN sales s ON e.emp_id = s.emp_id
ORDER BY e.name
''').show()
```

```python
spark.sql('''
WITH totals AS (
  SELECT emp_id, SUM(amount) AS total_amount
  FROM sales
  GROUP BY emp_id
  HAVING SUM(amount) >= 100
)
SELECT e.name, t.total_amount
FROM totals t
JOIN employees e ON e.emp_id = t.emp_id
''').show()
```

## Interview
- WHERE filters rows; HAVING filters groups.
- LEFT JOIN keeps unmatched left rows.
- Prefer CTEs for clarity.
