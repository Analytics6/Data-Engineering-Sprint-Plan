```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AzureDE-InterviewPrep')
    .master('local[*]')
    .config('spark.sql.shuffle.partitions', '4')
    .getOrCreate()
)
spark
```

```python
spark.createDataFrame([(1, 'East'), (2, 'West')], ['customer_key', 'region']).createOrReplaceTempView('dim_customer')
spark.createDataFrame([(1, 1, 20240101, 50.0)],
    ['sales_key', 'customer_key', 'date_key', 'amount']).createOrReplaceTempView('fact_sales')
```

```python
spark.sql('''
SELECT c.region, SUM(f.amount) AS revenue
FROM fact_sales f
JOIN dim_customer c ON f.customer_key = c.customer_key
GROUP BY c.region
''').show()
```
