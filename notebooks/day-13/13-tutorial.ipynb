# Day 13 Tutorial — ADF Control Flow & Orchestration

**Goal:** Lookup -> ForEach metadata-driven pattern + error handling.


## Key activities
Get Metadata, Lookup, ForEach, If/Switch, Until, Execute Pipeline

Dependency conditions: Succeeded / Failed / Completed / Skipped


```python
config = [
    {'src_table': 'dbo.Customers', 'sink_table': 'stg.customers', 'watermark_col': 'ModifiedDate'},
    {'src_table': 'dbo.Orders', 'sink_table': 'stg.orders', 'watermark_col': 'OrderDate'},
    {'src_table': 'dbo.Products', 'sink_table': 'stg.products', 'watermark_col': 'ModifiedDate'},
]
for row in config:
    print('Copy {0} -> {1} on {2}'.format(row['src_table'], row['sink_table'], row['watermark_col']))
```

## Parent-child pattern
Parent Lookup config -> ForEach -> Execute Pipeline child with parameters.
