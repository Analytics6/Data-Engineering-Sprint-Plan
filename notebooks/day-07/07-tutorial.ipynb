```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AzureDE-InterviewPrep')
    .master('local[*]')
    .config('spark.sql.shuffle.partitions', '4')
    .getOrCreate()
)
spark
```

```python
emp = spark.createDataFrame(
    [
        (1, 'Alice', 'Sales', 70000, '2020-01-01'),
        (2, 'Bob', 'Sales', 70000, '2019-06-01'),
        (3, 'Carol', 'Eng', 90000, '2021-03-01'),
        (4, 'Dan', 'Eng', 95000, '2018-01-01'),
        (5, 'Eve', 'Eng', 90000, '2022-01-01'),
    ],
    ['id', 'name', 'dept', 'salary', 'hire_date'],
)
w = Window.partitionBy('dept').orderBy(F.col('salary').desc(), F.col('hire_date').asc())
emp.withColumn('rn', F.row_number().over(w)).filter(F.col('rn') <= 2).show()
w_avg = Window.partitionBy('dept')
emp.withColumn('dept_avg', F.avg('salary').over(w_avg)).filter(F.col('salary') > F.col('dept_avg')).show()
```
