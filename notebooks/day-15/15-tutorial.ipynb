## Incremental patterns

3. CDC / change tracking

Performance levers: parallel copy, DIUs, partitioned reads, staging


```python
import pandas as pd
from datetime import datetime, timedelta
control = pd.DataFrame([{'pipeline': 'sales_copy', 'watermark': '2024-01-07 00:00:00'}])
source_rows = pd.DataFrame([
    {'id': 1, 'modified': '2024-01-06 10:00:00', 'amount': 10},
    {'id': 2, 'modified': '2024-01-07 12:00:00', 'amount': 20},
    {'id': 3, 'modified': '2024-01-08 09:00:00', 'amount': 30},
])
wm = control.loc[0, 'watermark']
source_rows[source_rows['modified'] > wm]
```

```python
def partitions_for_range(start, end):
    s = datetime.fromisoformat(start)
    e = datetime.fromisoformat(end)
    out = []
    cur = s
    while cur <= e:
        out.append(cur.strftime('raw/sales/%Y/%m/%d/'))
        cur += timedelta(days=1)
    return out

partitions_for_range('2024-01-01', '2024-01-03')
```
