```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AzureDE-InterviewPrep')
    .master('local[*]')
    .config('spark.sql.shuffle.partitions', '4')
    .getOrCreate()
)
spark
```

```python
spark.createDataFrame(
    [(1, 'c1', 100, '2024-01-01'), (2, 'c1', 200, '2024-01-02'), (3, 'c2', 50, '2024-01-01')],
    ['order_id', 'customer_id', 'amount', 'dt'],
).createOrReplaceTempView('orders')
```

```python
spark.sql('''
WITH daily AS (
  SELECT customer_id, dt, SUM(amount) AS amt
  FROM orders GROUP BY customer_id, dt
)
SELECT *, LAG(amt) OVER (PARTITION BY customer_id ORDER BY dt) AS prev_amt
FROM daily
''').show()
```
