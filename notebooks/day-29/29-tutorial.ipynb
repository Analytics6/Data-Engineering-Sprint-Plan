```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql import types as T
from pyspark.sql.window import Window

spark = (
    SparkSession.builder
    .appName('AzureDE-InterviewPrep')
    .master('local[*]')
    .config('spark.sql.shuffle.partitions', '4')
    .getOrCreate()
)
spark
```

## Drill 1 — latest record


```python
df = spark.createDataFrame(
    [('c1', '2024-01-01', 10), ('c1', '2024-01-03', 20), ('c2', '2024-01-02', 5)],
    ['customer_id', 'dt', 'amt'],
)
w = Window.partitionBy('customer_id').orderBy(F.col('dt').desc())
df.withColumn('rn', F.row_number().over(w)).filter('rn = 1').show()
```

## Drill 2 — running total


```python
df.createOrReplaceTempView('t')
```

```python
spark.sql('''
SELECT customer_id, dt, amt,
       SUM(amt) OVER (PARTITION BY customer_id ORDER BY dt) AS running_amt
FROM t
''').show()
```
